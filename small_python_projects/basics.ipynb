1. Simple calculator for two numbers 
    - Ask for two numbers and an operator, then print the result. Practice input, type conversion, and basic arithmetic.

```python
allowed_operations = ['+', '-', '*', '/', '^']
allowed_operations_str = ""
for op in allowed_operations:
    allowed_operations_str += op
    if op != allowed_operations[-1]:
        allowed_operations_str += " ; "

try:
    number1 = float(input(f"Enter left number of operation: "))
except ValueError:
    number1 = "NaN"
    print("Error: not a number.")


operation = input(f"Enter operation [{allowed_operations_str}]: ")
while operation not in allowed_operations:
    operation = input(f"Enter a valid operation [{allowed_operations_str}]: ")

if operation != '^':
    try:
        number2 = float(input(f"Enter right number of operation: "))
    except ValueError:
        number2 = "NaN"
        print("Error: not a number.")
else:
    try:
        number2 = int(input(f"Enter right whole number of operation: "))
    except ValueError:
        number2 = "NaN"
        print("Error: not a valid number.")

def add(a: float, b: float) -> float:
    if a == "NaN" or b == "NaN":
        return
    
    return a + b

def substract(a: float, b: float) -> float:
    if a == "NaN" or b == "NaN":
        return
    
    return a - b

def multiply(a: float, b: float) -> float:
    if a == "NaN" or b == "NaN":
        return
    
    return a * b

def divide(a: float, b: float) -> float:
    if a == "NaN" or b == "NaN":
        return
    
    if b == 0:
        return 
    
    return a / b

def square(a: float, b: int) -> float:
    if a == "NaN" or b == "NaN":
        return
    
    if a == 0:
        return 0.0

    if a == 1:
        return 1.0

    if b == 0:
        return 1.0

    if b < 0:
        return 

    result = 1.0
    for i in range(b):
        result *= a
    return result


def print_result(result: float):
    if result is None:  # None because i do just "return" in the functions -> returns None
        print(f"Math error: wrong number input - not a number / dividing by 0 / attempted square root.")
        return
    
    print(f"{number1} {operation} {float(number2)} = {result}")

if operation == '+':
    print_result(add(number1, number2))
elif operation == '-':
    print_result(substract(number1, number2))
elif operation == '*':
    print_result(multiply(number1, number2))
elif operation == '/':
    print_result(divide(number1, number2))
else:
    print_result(square(number1, number2))
    
```

2. Temperature converter (C/F/K)
    -  Convert between Celsius, Fahrenheit, and Kelvin. Focus on formulas and clean output formatting.

```python
def convert_temperature(temperature, type_input, type_result):
    # °F = (°C × 9/5) + 32
    if type_input == 'C' and type_result == 'F':
        return (temperature * 9 / 5) + 32 
    # K = °C + 273.15
    if type_input == 'C' and type_result == 'K':
        return temperature + 273.15
    # °C = (°F − 32) × 5/9
    if type_input == 'F' and type_result == 'C':
        return (temperature - 32) * 5 / 9
    #  K = (°F + 459.67) × 5/9
    if type_input == 'F' and type_result == 'K':
        return (temperature + 459.67) * 5 / 9
    # °C = K − 273.15
    if type_input == 'K' and type_result == 'C':
        return temperature - 273.15
    # °F = (K − 273.15) × 9/5 + 32
    if type_input == 'K' and type_result == 'F':
        return (temperature - 273.15) * 9 / 5 + 32
    return "Error: wrong input."

user_input = input("Enter temperature, the temperature type and the desired temperature type, ex. [ 30C->F ]")
temperature = float(user_input[0:-4])
temp_type_input = user_input[-4]
temp_type_result = user_input[-1]
print(f"{temperature}{temp_type_input} converted to {temp_type_result} = {convert_temperature(temperature, temp_type_input, temp_type_result)}{temp_type_result}")
```
