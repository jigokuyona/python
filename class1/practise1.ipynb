1. Napište program, který uživatele nejprve požádá o zadání celého čísla X a poté o zadání čísla Y. Pokud Y dělí číslo X beze zbytku vypíše "X je dělitelné Y beze zbytku" , jinak vypíše "Y nedělí X beze zbytku".

```python
X = int(input("input whole number: "))
Y = float(input("input any number: "))

if X % Y == 0:
    print(f"X [{X}] can be devided by Y [{Y}] without a remainder.")
else:
    print(f"Y [{Y}] cannot divide X [{X}] without a remainder.")
```

2. Napište program, který se nejprve zeptá na číslo X a Y a poté vrátí menší číslo, pokud jsou obě čísla lichá. Pokud je alespoň jedno číslo sudé, vrátí větší číslo. 

```python
X = float(input("input whole number: "))
Y = float(input("input any number: "))

if X % 2 == 1 and Y % 2 == 1:
    print(X) if X<=Y else print(Y)
else:
    print(X) if X>=Y else print(Y)
```

3. Vytvořte program, která vypočte kořeny kvadratické rovnice $ax^2 + bx + c = 0$. Program by měl řešit, kdy jsou řešením komplexní kořeny a kdy reálné kořeny. 

```python
a = int(input("input whole number for a: "))
b = int(input("input whole number for b: "))
c = int(input("input whole number for c: "))
D = b * b - 4 * a * c
if D < 0:
    print("Complex root of quadratic function.")
elif D > 0:
    x1 = (-b + D)/2 * a
    x2 = (-b - D)/2 * a
    print(f"x1 = {x1}, x2 = {x2}")
else:
    x = -b/2 * a
    print("x = ", x)
```

4. Napište program, který vytiskne následující čísla na 2 desetinné čísla. Hodnota proměnných se ale nezmění.  

```python
a = 2.3578
b = 3.1
def my_round(num: float, operator: int):
    string_full = str(num)[0:2] + str(num)[2:operator+2]
    if len(string_full) < operator+2:
        while len(string_full) < operator+2:
            string_full += "0" 
    return string_full

print(my_round(a, 2), my_round(b, 2))
```

5. Vytvořte program, který uživatele požádá o tři čísla a vrátí True pokud mohou tyto tři čísla tvořit strany trojúhelníka. Trojúhelník lze sestrojit pouze pokud jsou čísla kladná a suma jakýchkoliv dvou stran je větší než strana 3.   
Vstup: 5, 7, 1   
Výstup: False  


```python
a = int(input("input whole number for a: "))
b = int(input("input whole number for b: "))
c = int(input("input whole number for c: "))
def is_triangle(a: int, b: int, c: int) -> bool:
    if a > 0 and b > 0 and c > 0:
        if a + b > c and a + c > b and b + c > a:
            return True
    return False
print(f"[{a},{b},{c}] -> is triangle? = {is_triangle(a, b, c)}")
```

7. Vytvořte program, který uživatele požádá o tři čísla a vrátí největší z nich. 

```python
a = int(input("input whole number for a: "))
b = int(input("input whole number for b: "))
c = int(input("input whole number for c: "))
def biggest_num(a: int, b: int, c: int) -> int:
    return a if a > b and a > c else b if b > a and b > c else c
print(f"[{a},{b},{c}] -> biggest = {biggest_num(a, b, c)}")
```

8. Vytvořte program, který vypočte cenu za zboží. Cena zboží je závislá na počtu odebraných kusů.  
prvních 100 ks stojí 1,50Kč/kus  
dalších 100 ks stojí 1,30Kč/kus  
od 200 a dále  stojí 1,00Kč/kus

Vstup: 50  
Výstup: 75 Kč  
Vstup: 180  
Výstup: 254 Kč    

```python
amount = int(input("input whole number (amount): "))

def calculate_price(amount: int) -> float:
    if amount < 100:
        return amount * 1.5
    elif 100 <= amount < 200:
        return 150 + ((amount - 100) * 1.3)
    else:
        return 150 + 130 + amount - 200        
    
print(f"amount = {amount} -> price: {int(calculate_price(amount))} CZK")
```

9. Vytvořte program, který vrátí známku na základě počtu bodů ze semestrální práce.  
90 - 100 bodů: A  
80 -  90 bodů: B  
70 -  80 bodů: C  
60 -  70 bodů: D  
50 -  60 bodů: E  
méně         : F  

Vstup: 70  
Výstup: C 

```python
points = int(input("input amount of points (0-100): "))
while points < 0 or points > 100:
    points = int(input("input amount of points (0-100): "))

def calculate_grade(points: int) -> chr:
    grades = ['A', 'B', 'C', 'D', 'E', 'F']
    for i in range(len(grades)-1):
        if points >= 100 - ((i + 1) * 10):
            return grades[i]
    return grades[5]

print(f"points: {points} -> grade: {calculate_grade(points)}")
```

10. Vytvořte program, který uživatele požádá o měsíc v roce reprezentovaný celým číslem v rozsahu 1-12 a vrátí počet dní tohoto měsíce.

Výstup: 31  

```python
month = int(input("input a month number (1-12): "))
while month < 1 or month > 12:
    month = int(input("input a month number (1-12): "))

months_days = {
    1:31,
    2:28,
    3:31,
    4:30,
    5:31,
    6:30,
    7:31,
    8:31,
    9:30,
    10:31,
    11:30,
    12:31
}

def days_count_month(month: int) -> int:
    return months_days[month]

print(f"month: {month} -> days: {days_count_month(month)}")
```
